```julia
function outer(x, y, f)
    if (f == "*")
        res = x .* y'
    elseif (f == "+")
        res = x .+ y'
    elseif (f == "^")
        res = x .^ y'
    elseif (f == "-")
        res = x .- y'
    elseif (f == "%")
        res = x .% y'
    end
end
```

```julia
A1 = outer(0:4, 0:4, "+")
A1
```

```julia
A2 = outer(0:4, 1:5, "^")
A2
```

```julia
A3 = outer(outer(0:4, 0:4, "+"), 5, "%")
```

```julia
A4 = outer(outer(0:9, 0:9, "+"), 10, "%")
```
