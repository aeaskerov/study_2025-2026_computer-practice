```julia
function productive_matrix(matrix, size)
    ans = ""
    E = [1 0; 0 1]
    Y = rand(0:1000, size)
    S = E - matrix
    X = S\Y
    for i in 1:1:size
       for j in 1:1:size
            if X[i] < 0
                ans = false
                break
            else
                ans = true
            end
        end
    end
    return ans
end

A = [1 2; 3 4]
productive_matrix(A, 2)
```

```julia
B = 1/2 * A
productive_matrix(B, 2)
```

```julia
C = 1/10 * A
productive_matrix(C, 2)
```

```julia
function productive_matrix_2(matrix, size)
    ans = ""
    E = [1 0; 0 1]
    new_matrix = E - matrix
    inv_new_matrix = inv(new_matrix)
    for i in 1:1:size
       for j in 1:1:size
            if inv_new_matrix[i, j] < 0
                ans = false
                break
            else
                ans = true
            end
        end
    end
    return ans
end

A = [1 2; 3 1]
productive_matrix_2(A, 2)
```

```julia
B = 1/2 * A
productive_matrix_2(B, 2)
```

```julia
C = 1/10 * A
productive_matrix_2(C, 2)
```

```julia
function productive_matrix_3(matrix, size)
    ans = ""
    eigenvalues = eigvals(matrix)
    for i in 1:1:size
        if abs(eigenvalues[i]) >= 1
            ans = false
            break
        else
            ans = true
        end
    end
    return ans
end

A = [1 2; 3 1]
productive_matrix_3(A, 2)
```

```julia
B = 1/2 * A
productive_matrix_3(B, 2)
```

```julia
C = 1/10 * A
productive_matrix_3(C, 2)
```

```julia
D = [0.1 0.2 0.3; 0 0.1 0.2; 0 0.1 0.3]
productive_matrix_3(D, 2)
```
