```julia
# подключаем необходимые пакеты
import Pkg
Pkg.add("DifferentialEquations")
using DifferentialEquations
```

```julia
# задаём описание модели с начальными условиями
a = 0.98
f(u,p,t) = a*u
u0 = 1.0
# задаём интервал времени
tspan = (0.0, 1.0)
# решение
prob = ODEProblem(f,u0,tspan)
sol = solve(prob)
```

```julia
# подключаем необходимые пакеты
# Pkg.add("Plots")
using Plots
# строим графики
plot(sol, linewidth=5, title="Модель экспоненциального роста",
    xaxis="Время", yaxis="u(t)", label="Численное решение")
plot!(sol.t, t->1.0*exp(a*t), lw=3, ls = :dash,
    label="Аналитическое решение")
```

```julia
# задаём точность решения
sol = solve(prob, abstol=1e-8, reltol=1e-8)
println(sol)
```

```julia
# строим график
plot(sol, lw=2, color="black",
    title="Модель экспоненциального роста", xaxis="Время",
    yaxis="u(t)", label="Численное решение")
plot!(sol.t, t->1.0*exp(a*t), lw=3, ls=:dash, color="red",
    label="Аналитическое решение")
```

```julia
# подключаем необходимые пакеты
# import Pkg
# Pkg.add("DifferentialEquations")
using DifferentialEquations, Plots;
# задаём описание модели
function lorenz!(du,u,p,t)
    sigma, rho, beta = p
    du[1] = sigma*(u[2]-u[1])
    du[2] = u[1]*(rho-u[3]) - u[2]
    du[3] = u[1]*u[2] - beta*u[3]
end
# задаём начальное условие
u0 = [1.0, 0.0, 0.0]
# задаём значения параметров
p = (10, 28, 8/3)
# задаём интервал времени
tspan = (0.0, 100.0)
# решение
prob = ODEProblem(lorenz!,u0,tspan,p)
sol = solve(prob)
```

```julia
# подключаем необходимые пакеты
# Pkg.add("Plots")
using Plots
# строим график
plot(sol, vars=(1,2,3), lw=2, title="Аттрактор Лоренца",
    xaxis="x", yaxis="y", zaxis="z", legend=false)
```

```julia
# отключаем интерполяцию
plot(sol, vars=(1,2,3), denseplot=false, lw=1,
    title="Аттрактор Лоренца", xaxis="x", yaxis="y", zaxis="z",
    legend=false)
```

```julia
# Задание 3
function SIR(u,p,t)
    (S,I,R) = u
    (β, ν) = p
    N = S + I + R
    dS = -(β*I*S)/N
    dI = (β*I*S)/N - ν*I
    dR = ν*I
    return [dS, dI, dR]
end
δt = 0.1
tmax = 100.0
tspan = (0.0, tmax)
u0 = [555.0, 10.0, 0.0] # S, I, R
p = [0.2, 0.1] # β, ν
prob_ode = ODEProblem(SIR, u0, tspan, p)
sol_ode = solve(prob_ode, dt=δt)
```

```julia
plot(sol_ode, title="SIR-модель", label=["S" "I" "R"], lw=3)
```

```julia
k = u0[1] + 50
anim = @animate for t in 0.0:δt:tmax
    idx = sol_ode.t .<= t
    plot(sol_ode.t[idx], getindex.(sol_ode.u[idx], 1),
         label="S", lw=2, xlabel="Время", ylabel="Численность",
         xlim=(0,tmax), ylim=(0, k), title="SIR-модель")
    plot!(sol_ode.t[idx], getindex.(sol_ode.u[idx], 2),
        label="I", lw=2)
    plot!(sol_ode.t[idx], getindex.(sol_ode.u[idx], 3),
        label="R", lw=2)
end

gif(anim, "./sir_animation.gif", fps=30)
```

```julia
# Задание 4
function SEIR(u,p,t)
    (S,E,I,R) = u
    (β, γ, δ) = p
    N = S + E + I + R
    dS = -(β*I*S)/N
    dE = (β*S*I)/N - δ*E
    dI = δ*E - γ*I
    dR = γ*I
    return [dS, dE, dI, dR]
end
δt = 0.1
tmax = 100.0
tspan = (0.0, tmax)
u0 = [555.0, 10.0, 10.0, 0.0]
p = [0.2, 0.1, 0.3]
prob_ode = ODEProblem(SEIR, u0, tspan, p)
sol_ode = solve(prob_ode, dt=δt)
```

```julia
plot(sol_ode, title="SEIR-модель", label=["S" "E" "I" "R"], lw=3)
```

```julia
# Задание 5
using Plots
a, c, d = 2,1,5
t_max = 50
X1_0 = 0.5
X2_0 = 0.3
X1 = zeros(t_max)
X2 = zeros(t_max)
X1[1] = X1_0
X2[1] = X2_0
for t in 1:t_max-1
    X1[t+1] = a*X1[t]*(1-X1[t])-X1[t]*X2[t]
    X2[t+1] = -c*X2[t]+d*X1[t]*X2[t]
end
plot(1:t_max, X1, label="Жертвы (X1)", title="Численное решение",
    xlabel="Время", ylabel="Численность", lw=3)
plot!(1:t_max, X2, label="Хищники (X2)", lw=3)
```

```julia
# Задание 6
function competition(u,p,t)
    (x, y) = u
    α, β = p
    dx = α*x - β*x*y
    dy = α*y - β*x*y
    return [dx, dy]
end
tspan = (0.0, 6.0)
u0 = [10, 5]
p = [1, 0.01]
prob = ODEProblem(competition, u0, tspan, p)
sol = solve(prob, Rodas5())
```

```julia
plot(sol, title="Конкурентные отношения", label=["x" "y"], lw=3)
```

```julia
plot(sol, vars=(1,2), title="Конкурсные отношения", lw=3)
```

```julia
animate(sol, "./competitive_relationship.gif",
    title="Конкурсные отношения", label=["x" "y"], lw=3)
```

```julia
# Задание 7
function harmonic(u,p,t)
    x, y = u
    γ, ω = p
    dx = y
    dy = -2*γ*y - ω^2*x
    return [dx, dy]
end
tspan = (0.0, 2*pi)
u0 = [-0.5, 0]
p = [0, 5]
prob = ODEProblem(harmonic, u0, tspan, p)
sol = solve(prob, Tsit5())
```

```julia
plot(sol, title="Консервативный гармонический осциллятор", lw=3)
```

```julia
plot(sol, vars=(1,2), lw=3)
```

```julia
animate(sol, title="Консервативный гармонический осциллятор", lw=3)
```

```julia
# Задание 8
function harmonic2(du, u , p, t)
    γ, ω0 = p
    du[1] = u[2]
    du[2] = -2*γ*u[2]-ω0^2*u[1]
end
u0 = [1.0, 0.0]
γ = 0.1
ω0 = 1.0
p = (γ, ω0)
tspan = (0.0, 70.0)
prob = ODEProblem(harmonic2, u0, tspan, p)
sol = solve(prob, Tsit5())
```

```julia
plot(sol, title="Свободные колебания гармонического осциллятора",
    lw=3)
```
